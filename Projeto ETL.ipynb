```python
import pandas as pd
import numpy as np
```

```python
df_vendas =pd.read_csv("vendas_tech.csv", low_memory=False)
df_gerentes = pd.read_excel("gerentes_lojas.xlsx")

display(df_vendas)
display(df_gerentes)
```

```python
#Inspeção dos dados
display(df_vendas.head(15))
display(df_vendas.tail(15))
display(df_vendas.sample(15))
display(df_vendas.shape)
display(list(df_vendas.columns))
display(df_vendas.info())
display(df_vendas.describe())
```

```python
# tratamento de dados
display(df_vendas['Loja'])
display(df_vendas[['Loja','Cliente']])
#colunas
df_analise = df_vendas.drop(columns=['Data_Base'])
#nulos
# df_analise = df_analise.dropna() # EXCLUIR TODAS AS LINHAS COM VALORES NULOS E VAZIOS
df_analise['Loja'] = df_analise['Loja'].fillna('Online') # SUBSTITUIR VALORES NULOS E VAZIOS
#tipos de dados
df_analise['Data'] = pd.to_datetime(df_analise['Data'], format='%Y-%m-%d')
#padronização
df_analise['Loja'] = df_analise['Loja'].str.strip()
df_analise['Loja'] = df_analise['Loja'].str.title()
#Duplicatas
df_analise = df_analise.drop_duplicates(subset=['ID_Pedido'])
```

```python
#Criar novas colunas

#Faturamento
df_analise['Faturamento'] = df_analise['Qtd'] * df_analise['Preco_Unitario']
#Região
dicionario_regiao = {
    'São Paulo': 'Sudeste',
    'Belo Horizonte': 'Sudeste',
    'Online': 'Online',
    'Rio De Janeiro': 'Sudeste',
    'Salvador': 'Nordeste',
    'Recife': 'Nordeste',
    'Curitiba': 'Sul',
    'Porto Alegre': 'Sul'
}
df_analise['Regiao'] = df_analise['Loja'].map(dicionario_regiao)
#forma de vendas
df_analise['Forma_de_Venda'] = np.where(df_analise['Loja'] == 'Online', 'Online', 'Presencial')
```

```python
display(df_analise)
display(df_analise.isna().sum())
```

```python
# Analisar - > filtrar
df_analise = df_analise.sort_values(by=['Data', 'Faturamento']).reset_index(drop=True)
#.loc: id do pedido -> por nome da coluna e por nome da linha
id_pedido = 4
loja = df_analise.loc[3, 'Loja']
produto = df_analise.loc[3, 'Produto']
cliente = df_analise.loc[3, 'Cliente']
print(f'id do pedido: {id_pedido}, Loja: {loja}, Produto: {produto}, Cliente: {cliente}')
#.iloc -> por posição
loja = df_analise.iloc[3, 2]
produto = df_analise.iloc[3, 3]
cliente = df_analise.iloc[3, 6]
print(f'id do pedido: {id_pedido}, Loja: {loja}, Produto: {produto}, Cliente: {cliente}')
#condicional
df_id_pedido_4 = df_analise[df_analise['ID_Pedido'] == 4]
display(df_id_pedido_4)
#Exportar pedaços da base
df_vendas_sp = df_analise[df_analise['Loja'] == 'São Paulo']
df_vendas_sp.to_csv('vendas_sp.csv', index=False)
display(df_vendas_sp)

#exportar vendas 2024
df_vendas_2024 = df_analise[df_analise['Data'].dt.year == 2024]
#df_vendas_2024 = df_analise[df_analise['Data'] >= '2024-01-01']
df_vendas_2024.to_csv('vendas_2024.csv', index=False)
display(df_vendas_2024)

#duplas condições
df_vendas_HDMI_SUL = df_analise[(df_analise['Regiao'] == 'Sul') & (df_analise['Produto'] == 'Cabo HDMI')]
display(df_vendas_HDMI_SUL)
```

```python
#Analise por agrupamento
analise_lojas = df_analise[['Loja', 'Faturamento']].groupby('Loja').sum().sort_values(by='Faturamento', ascending=False).reset_index()
analise_lojas['Faturamento'] = analise_lojas['Faturamento'].map('R${:,.2f}'.format)
display(analise_lojas)

#Ranking de produtos que mais vederam no online
df_vendas_online = df_analise[df_analise['Loja'] == 'Online']
analise_produtos_online = df_vendas_online[['Produto', 'Faturamento']].groupby('Produto').sum().sort_values(by='Faturamento', ascending=False).reset_index()
display(analise_produtos_online)

#Ranking de produtos que mais vederam no online
df_vendas_online = df_analise[df_analise['Loja'] == 'Online']
analise_produtos_online_qtd = df_vendas_online[['Produto', 'Qtd']].groupby('Produto').sum().sort_values(by='Qtd', ascending=False).reset_index()
analise_produtos_online_qtd = analise_produtos_online_qtd.rename(columns={'Qtd': 'Vendas Totais'})
display(analise_produtos_online_qtd)

#Ranking por loja e produto
analise_produtos_em_lojas = df_analise[['Loja','Produto', 'Qtd']].groupby(['Loja','Produto']).sum()
analise_lojas_em_produtos = df_analise[['Produto','Loja', 'Qtd']].groupby(['Produto','Loja']).sum()
with pd.option_context('display.max_rows', None):
    display(analise_produtos_em_lojas)
    display(analise_lojas_em_produtos)
```
